```python
dfs = []

for yr in range(2015, 2027):
    try:
        print(f"{yr}年のデータを取得中...")

        df_yr = pitching_stats(yr, yr, qual=20)

        if df_yr is not None and not df_yr.empty:
            if 'Season' not in df_yr.columns:
                df_yr['Season'] = yr

            dfs.append(df_yr)

            print(f"  成功: {len(df_yr)} 件")

        else:
            print(f"  データなし: {yr}年")

    except Exception as e:
        print(f"  エラー: {yr}年 / {e}")
```
