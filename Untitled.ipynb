```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from pybaseball import pitching_stats
```

```python
tj_database_url = "https://docs.google.com/spreadsheets/d/1gQujXQQGOVNaiuwSN680Hq-FDVsCwvN-3AazykOBON0/export?format=csv"
```

```python
raw_preview = pd.read_csv(tj_database_url, nrows=10, header=None)

header_idx = 0
for idx, row in raw_preview.iterrows():
    row_str = " ".join(row.dropna().astype(str)).lower()
    if ('player' in row_str or 'name' in row_str) and ('date' in row_str or 'tj' in row_str or 'surgery' in row_str):
        header_idx = idx
        break
```

```python
tj_raw_df = pd.read_csv(tj_database_url, header=header_idx)
print(f"ヘッダー検出行: Row {header_idx}")
print(f"取得データ件数: 全 {len(tj_raw_df)} 件")
```

```python
player_col = None
date_col = None

for col in tj_raw_df.columns:
    c_lower = str(col).lower()
    if not player_col and any(k in c_lower for k in ['player', 'name', '選手']):
        player_col = col
    if not date_col and any(k in c_lower for k in ['date', 'surgery', 'tj', '日付', '手術']):
        date_col = col
        
if not player_col:
    player_col = tj_raw_df.columns[0]
if not date_col:
    date_cols = [c for c in tj_raw_df.columns if any(k in str(c).lower() for k in ['date', 'surg', 'year', 'day'])]
    date_col = date_cols[0] if date_cols else tj_raw_df.columns[3]
```

```python
print(f"対象カラム特定 - 選手名列: '{player_col}', 手術日列: '{date_col}'")
```

```python
tj_clean = tj_raw_df[[player_col, date_col]].dropna().copy()
tj_clean.columns = ['Name', 'TJ_Date']
```

```python
tj_clean['TJ_Date'] = pd.to_datetime(tj_clean['TJ_Date'], errors='coerce')
tj_clean = tj_clean.dropna(subset=['TJ_Date'])
tj_clean['TJ_Year'] = tj_clean['TJ_Date'].dt.year.astype(int)
```

```python
tj_players_df = tj_clean[(tj_clean['TJ_Year'] >= 2015) & (tj_clean['TJ_Year'] <= 2026)].drop_duplicates(subset=['Name', 'TJ_Year'])
print(f"分析対象期間 (2015-2026) のTJ手術件数: {len(tj_players_df)} 件")
```

```python
df_mlb_raw = pitching_stats(2015, 2026, qual=30)
```

```python
if isinstance(df_mlb_raw.columns, pd.MultiIndex):
    df_mlb_raw.columns = ['_'.join([str(e) for e in col if str(e)]).strip() for col in df_mlb_raw.columns]

df_mlb_raw.columns = [str(c).strip() for c in df_mlb_raw.columns]
raw_cols = list(df_mlb_raw.columns)
```

```python
name_col = None
for c in raw_cols:
    c_lower = c.lower()
    if 'name' in c_lower or 'player' in c_lower:
        name_col = c
        break
if not name_col:
    name_col = raw_cols[0]

target_cols = {'Name': name_col}

def find_col(keywords):
    for c in raw_cols:
        c_lower = c.lower()
        if any(k in c_lower for k in keywords):
            return c
    return None
```

```python
season_col = find_col(['season', 'year'])
era_col = find_col(['era'])
fip_col = find_col(['fip'])
k9_col = find_col(['k/9', 'k_9', 'so/9'])
bb9_col = find_col(['bb/9', 'bb_9'])
fbv_col = find_col(['fbv', 'vfa'])
```

```python
if season_col: target_cols['Season'] = season_col
if era_col: target_cols['ERA'] = era_col
if fip_col: target_cols['FIP'] = fip_col
if k9_col: target_cols['K/9'] = k9_col
if bb9_col: target_cols['BB/9'] = bb9_col
if fbv_col: target_cols['FBv'] = fbv_col
```

```python
rename_dict = {v: k for k, v in target_cols.items()}
df_mlb = df_mlb_raw[list(target_cols.values())].rename(columns=rename_dict).copy()
```

```python
if 'Name' in df_mlb.columns:
    df_mlb['Name'] = df_mlb['Name'].astype(str).str.strip()
if 'Name' in tj_players_df.columns:
    tj_players_df['Name'] = tj_players_df['Name'].astype(str).str.strip()

for col in ['FBv', 'FIP', 'K/9', 'BB/9', 'ERA']:
    if col in df_mlb.columns:
        df_mlb[col] = pd.to_numeric(df_mlb[col], errors='coerce')

print("抽出成功カラム:", df_mlb.columns.tolist())
```

```python
analysis_results = []

for idx, row in tj_players_df.iterrows():
    p_name = row['Name']
    tj_year = row['TJ_Year']

    pre_df = df_mlb[(df_mlb['Name'] == p_name) & (df_mlb['Season'] >= tj_year - 2) & (df_mlb['Season'] < tj_year)]
    post_df = df_mlb[(df_mlb['Name'] == p_name) & (df_mlb['Season'] > tj_year) & (df_mlb['Season'] <= tj_year + 2)]
    
    if not pre_df.empty and not post_df.empty:
        pre_fbv, post_fbv = pre_df['FBv'].mean(), post_df['FBv'].mean()
        pre_fip, post_fip = pre_df['FIP'].mean(), post_df['FIP'].mean()
        pre_k9, post_k9 = pre_df['K/9'].mean(), post_df['K/9'].mean()
        pre_bb9, post_bb9 = pre_df['BB/9'].mean(), post_df['BB/9'].mean()
        
        if not np.isnan(pre_fbv) and not np.isnan(post_fbv):
            analysis_results.append({
                'Name': p_name,
                'TJ_Year': tj_year,
                'Pre_FBv': pre_fbv,
                'Post_FBv': post_fbv,
                'Delta_FBv': post_fbv - pre_fbv,
                'Delta_FIP': post_fip - pre_fip,
                'Delta_K9': post_k9 - pre_k9,
                'Delta_BB9': post_bb9 - pre_bb9
            })

df_analysis = pd.DataFrame(analysis_results)
df_analysis.to_csv('tj_web_analysis_output.csv', index=False)
print(f"\n突合完了: 成績データが存在する復帰投手サンプル数: {len(df_analysis)} 名")
```
