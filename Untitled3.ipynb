```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pybaseball import chadwick_register, statcast_pitcher
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 300
plt.rcParams['font.size'] = 10
```

```python
tjs_url = "https://docs.google.com/spreadsheets/d/1gQujXQQGOVNaiuwSN680Hq-FDVsCwvN-3AazykOBON0/export?format=csv&gid=0"
```

```python
df_raw_all = pd.read_csv(tjs_url, header=None, low_memory=False)
```

```python
header_idx = None
for idx, row in df_raw_all.iterrows():
    row_vals = [str(v).lower() for v in row.values]
    if any('player' in v for v in row_vals) and any('date' in v or 'surgery' in v for v in row_vals):
        header_idx = idx
        break
```

```python
raw_tjs_df = pd.read_csv(tjs_url, skiprows=header_idx)
```

```python
def match_column(df, keywords):
    for kw in keywords:
        for col in df.columns:
            if kw.lower() in str(col).lower():
                return col
    raise KeyError(f"該当列なし: {keywords}")
```

```python
player_col = match_column(raw_tjs_df, ['player', 'name'])
date_col = match_column(raw_tjs_df, ['tj surgery date', 'date', 'surgery'])
level_col = match_column(raw_tjs_df, ['level', 'league'])
```

```python
raw_tjs_df['Surgery_Date'] = pd.to_datetime(raw_tjs_df[date_col], errors='coerce')
raw_tjs_df['TJS_Year'] = raw_tjs_df['Surgery_Date'].dt.year
```

```python
mlb_tjs = raw_tjs_df[
    (raw_tjs_df[level_col].astype(str).str.upper() == 'MLB') & 
    (raw_tjs_df['TJS_Year'] >= 2015) & 
    (raw_tjs_df['TJS_Year'] <= 2026)
].copy()
```

```python
cw_reg = chadwick_register()

mlb_tjs['name_lower'] = mlb_tjs[player_col].str.lower().str.strip()
cw_reg['name_lower'] = (cw_reg['name_first'] + ' ' + cw_reg['name_last']).str.lower().str.strip()

merged_targets = pd.merge(
    mlb_tjs, 
    cw_reg[['name_lower', 'key_mlbam']].drop_duplicates(subset=['name_lower']), 
    on='name_lower', 
    how='inner'
).dropna(subset=['key_mlbam'])

merged_targets['key_mlbam'] = merged_targets['key_mlbam'].astype(int)
```

```python
sample_targets = merged_targets.head(10)

compiled_records = []
all_pitch_logs = []

for idx, row in sample_targets.iterrows():
    p_name = row[player_col]
    p_id = row['key_mlbam']
    t_year = int(row['TJS_Year'])
    
    try:
        df_sp = statcast_pitcher(start_dt="2015-03-31", end_dt="2026-11-01", player_id=p_id)
        if df_sp.empty:
            continue
            
        df_sp['game_date'] = pd.to_datetime(df_sp['game_date'])
        df_sp['year'] = df_sp['game_date'].dt.year
        df_sp['player_name'] = p_name
        all_pitch_logs.append(df_sp)
        
        for yr, group in df_sp.groupby('year'):
            if len(group) < 80:
                continue
                
            ff_df = group[group['pitch_type'] == 'FF']
            ff_velo = ff_df['release_speed'].mean() if not ff_df.empty else np.nan
            
            swings = group[group['description'].isin([
                'swinging_strike', 'swinging_strike_blocked', 'foul', 'foul_tip', 
                'hit_into_play', 'hit_into_play_score'
            ])]
            whiffs = group[group['description'].isin(['swinging_strike', 'swinging_strike_blocked'])]
            whiff_rate = (len(whiffs) / len(swings) * 100) if len(swings) > 0 else np.nan
            
            phase = None
            if yr == t_year - 1:
                phase = "Pre-TJS"
            elif yr == t_year + 1:
                phase = "Post-Year1"
            elif yr == t_year + 2:
                phase = "Post-Year2"
                
            if phase:
                compiled_records.append({
                    "Name": p_name,
                    "TJS_Year": t_year,
                    "Season": yr,
                    "Phase": phase,
                    "FF_Velo": ff_velo,
                    "Whiff_Rate": whiff_rate
                })
    except Exception as err:
        print(f"スキップ ({p_name}): {err}")
        continue
```

```python
df_analysis = pd.DataFrame(compiled_records)
```

```python
print("df_analysis の生成完了。件数:", len(df_analysis))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=df_analysis, x='Phase', y='FF_Velo', ax=axes[0], palette='Blues', width=0.4, boxprops=dict(alpha=0.7))
sns.stripplot(data=df_analysis, x='Phase', y='FF_Velo', ax=axes[0], color='black', alpha=0.6, jitter=0.2, size=6)
axes[0].set_title('Fastball Velocity Distribution by Phase', fontsize=11, fontweight='bold')
axes[0].set_ylabel('4-Seam Velocity (mph)')
axes[0].set_xlabel('Period relative to TJS')

sns.boxplot(data=df_analysis, x='Phase', y='Whiff_Rate', ax=axes[1], palette='Oranges', width=0.4, boxprops=dict(alpha=0.7))
sns.stripplot(data=df_analysis, x='Phase', y='Whiff_Rate', ax=axes[1], color='black', alpha=0.6, jitter=0.2, size=6)
axes[1].set_title('Whiff Percentage Distribution by Phase', fontsize=11, fontweight='bold')
axes[1].set_ylabel('Whiff Rate (%)')
axes[1].set_xlabel('Period relative to TJS')

plt.tight_layout()
plt.savefig('fig1_phase_distribution.png', dpi=300)
plt.show()
```

```python
pivoted = df_analysis.pivot(index='Name', columns='Phase', values=['FF_Velo', 'Whiff_Rate'])
pivoted.columns = ['_'.join(c) for c in pivoted.columns]

if 'FF_Velo_Pre-TJS' in pivoted.columns and 'FF_Velo_Post-Year1' in pivoted.columns:
    pivoted_clean = pivoted.dropna(subset=['FF_Velo_Pre-TJS', 'FF_Velo_Post-Year1']).copy()
    pivoted_clean['Velo_Diff1'] = pivoted_clean['FF_Velo_Post-Year1'] - pivoted_clean['FF_Velo_Pre-TJS']

    if not pivoted_clean.empty:
        plt.figure(figsize=(6.5, 5.0))
        ax = sns.regplot(
            data=pivoted_clean, 
            x='FF_Velo_Pre-TJS', 
            y='Velo_Diff1', 
            color='#2b5c8f',
            scatter_kws={'s': 50, 'alpha': 0.8},
            line_kws={'linewidth': 2}
        )

        for name_idx, row in pivoted_clean.iterrows():
            ax.annotate(
                str(name_idx).split()[-1],
                (row['FF_Velo_Pre-TJS'], row['Velo_Diff1']),
                textcoords="offset points",
                xytext=(5, 5),
                ha='left',
                fontsize=8
            )

        plt.title('Pre-TJS Velocity vs Post-Year1 Velocity Change', fontsize=11, fontweight='bold')
        plt.xlabel('Pre-TJS 4-Seam Velocity (mph)')
        plt.ylabel('Post-Year1 Velocity Change (mph)')
        plt.axhline(0, color='red', linestyle='--', alpha=0.5)
        plt.tight_layout()
        plt.savefig('fig2_velocity_correlation.png', dpi=300)
        plt.show()
```

```python
if all_pitch_logs:
    full_sp_df = pd.concat(all_pitch_logs, ignore_index=True)
    df_sp_filtered = full_sp_df[full_sp_df['pitch_type'].isin(['FF', 'SL', 'CH', 'CU', 'SI', 'FC'])].copy()
    
    pitch_counts = pd.crosstab(df_sp_filtered['year'], df_sp_filtered['pitch_type'], normalize='index') * 100
    
    ax = pitch_counts.plot(kind='bar', stacked=True, figsize=(8, 4.5), colormap='tab10')
    plt.title('Pitch Mix Transition Across Seasons', fontsize=11, fontweight='bold')
    plt.ylabel('Percentage of Total Pitches (%)')
    plt.xlabel('Season')
    plt.legend(title='Pitch Type', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig('fig3_pitch_mix.png', dpi=300)
    plt.show()
```
