```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pybaseball import chadwick_register, statcast_pitcher
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 300
```

```python
tjs_url = "https://docs.google.com/spreadsheets/d/1gQujXQQGOVNaiuwSN680Hq-FDVsCwvN-3AazykOBON0/export?format=csv&gid=0"
```

```python
df_raw_all = pd.read_csv(tjs_url, header=None, low_memory=False)
```

```python
header_idx = None
for idx, row in df_raw_all.iterrows():
    row_vals = [str(v).lower() for v in row.values]
    if any('player' in v for v in row_vals) and any('date' in v or 'surgery' in v for v in row_vals):
        header_idx = idx
        break
```

```python
raw_tjs_df = pd.read_csv(tjs_url, skiprows=header_idx)
```

```python
def match_column(df, keywords):
    for kw in keywords:
        for col in df.columns:
            if kw.lower() in str(col).lower():
                return col
    raise KeyError(f"該当する列が見つかりません: {keywords}. 存在する列名: {list(df.columns)}")
```

```python
player_col = match_column(raw_tjs_df, ['player', 'name'])
date_col = match_column(raw_tjs_df, ['tj surgery date', 'date', 'surgery'])
level_col = match_column(raw_tjs_df, ['level', 'league'])
```

```python
print(f"検出された列名 -> 選手名: '{player_col}', 手術日: '{date_col}', レベル: '{level_col}'")
```

```python
raw_tjs_df['Surgery_Date'] = pd.to_datetime(raw_tjs_df[date_col], errors='coerce')
raw_tjs_df['TJS_Year'] = raw_tjs_df['Surgery_Date'].dt.year
```

```python
mlb_tjs = raw_tjs_df[
    (raw_tjs_df[level_col].astype(str).str.upper() == 'MLB') & 
    (raw_tjs_df['TJS_Year'] >= 2015) & 
    (raw_tjs_df['TJS_Year'] <= 2026)
].copy()
```

```python
print(f"分析対象条件に合致したMLB投手数: {len(mlb_tjs)} 名")
```
